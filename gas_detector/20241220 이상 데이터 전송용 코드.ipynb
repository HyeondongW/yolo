```python
# 대구정밀 데이터 수집부를 종료하고 실행하십시오.

import datetime
import csv

from kafka import KafkaProducer
import sys, types
from json import dumps

import pandas as pd
import time

# publish 필요 코드
m = types.ModuleType('kafka.vendor.six.moves', 'Mock module')
setattr(m, 'range', range)
sys.modules['kafka.vendor.six.moves'] = m
bootstrap_servers = ['piai_kafka2.aiot.town:9092',]


df = pd.read_csv('test_sample.csv')

while True:  

    # 현재 시간을 UTC로 얻기
    current_time_utc = datetime.datetime.utcnow()

    # timestamp 생성 (UNIX epoch 시간 - 초 단위)
    timestamp = int(current_time_utc.timestamp())

    json_producer = KafkaProducer(    
        bootstrap_servers=bootstrap_servers, # 전달하고자 하는 카프카 브로커의 주소 리스트
        api_version=(2,2,2),
        value_serializer=lambda x:dumps(x).encode('utf-8') # 메시지의 값 직렬화
    )

    for index, row in df.iterrows():
        publish_dict = {
            "Humidity" : row['Humidity'],
            "Temperature" : row['Temprature'],
            "Gas(MQ-2)" : row['Gas(MQ-2)'],
            "Gas(MQ-4)" : row['Gas(MQ-4)'],
            "Timestamp" : timestamp
        }
        time.sleep(0.25)  # 1초 대기  
        
        # print(publish_dict)
        publish_result = json_producer.send('DGSP-SENSOR-DATA-SM2', value = publish_dict)

    print('cycle complete')
```
